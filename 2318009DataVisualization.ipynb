# **1. Import Libraries**

```python
# Import library
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
# Langkah 1: Mount Google Drive
from google.colab import drive
drive.mount('/content/drive')
```

# **2. Baca Dataset tip restoran**

```python
# Baca file .xlsx
from google.colab import drive
drive.mount('/content/drive')

file_path = '/content/drive/My Drive/Data Mining/tip_cleaned.csv'
df = pd.read_csv(file_path)
```

```python
# Tampilkan 5 data teratas
print(df.head())
```

```python
# Atur style visualisasi
sns.set(style="whitegrid")
```

# 3. melakukan Visualisasi distribusi total bill

```python
# 1️⃣ Visualisasi distribusi total bill
plt.figure(figsize=(8, 5))
sns.histplot(df['total_bill'], bins=20, kde=True, color='skyblue')
plt.title('Distribusi Total Bill')
plt.xlabel('Total Bill')
plt.ylabel('Frekuensi')
plt.show()
```

# 4.Melakukan visualisasi Rata-rata tip berdasarkan hari

```python
# 2️⃣ Rata-rata tip berdasarkan hari
plt.figure(figsize=(8, 5))
sns.barplot(x='day', y='tip', data=df, estimator='mean', palette='viridis')
plt.title('Rata-rata Tip Berdasarkan Hari')
plt.xlabel('Hari')
plt.ylabel('Rata-rata Tip')
plt.show()
```

# 5. Visualisasi Hubungan antara total bill dan tip

```python
# 3️⃣ Hubungan antara total bill dan tip
plt.figure(figsize=(8, 5))
sns.scatterplot(x='total_bill', y='tip', data=df, hue='sex', style='smoker', palette='cool')
plt.title('Hubungan Total Bill dan Tip')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.show()
```

# 6. Visualisasi Boxplot tip berdasarkan hari

```python
# 4️⃣ Boxplot tip berdasarkan hari
plt.figure(figsize=(8, 5))
sns.boxplot(x='day', y='tip', data=df, palette='Set2')
plt.title('Variasi Tip Berdasarkan Hari')
plt.xlabel('Hari')
plt.ylabel('Tip')
plt.show()
```

```python
# 5️⃣ Heatmap Korelasi antar Variabel Numerik

# Pilih hanya kolom numerik
numeric_df = df.select_dtypes(include=['float64', 'int64'])

# Hitung korelasi
corr = numeric_df.corr()

# Plot heatmap
plt.figure(figsize=(6, 4))
sns.heatmap(corr, annot=True, cmap='coolwarm', center=0)
plt.title('Heatmap Korelasi antar Variabel Numerik', fontsize=14)
plt.show()
```
